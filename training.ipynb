```python
train_df = train_df.sort_values(
    ["unit", "cycle"]
).reset_index(drop=True)
```

```python
max_cycle = train_df.groupby("unit")["cycle"].transform("max")

train_df["RUL"] = max_cycle - train_df["cycle"]
```

```python
FAILURE_HORIZON = 30

train_df["failure_risk"] = (
    train_df["RUL"] <= FAILURE_HORIZON
).astype(int)
```

```python
print(train_df["failure_risk"].value_counts())
```

```python
sensor_cols = [f"sensor_{i}" for i in range(1, 22)]

window = 10

for sensor in sensor_cols:
    # Change from previous cycle
    train_df[f"{sensor}_diff"] = (
        train_df.groupby("unit")[sensor].diff()
    )

    # Recent average
    train_df[f"{sensor}_roll_mean"] = (
        train_df.groupby("unit")[sensor]
        .transform(lambda x: x.rolling(window, min_periods=1).mean())
    )

    # Recent variation
    train_df[f"{sensor}_roll_std"] = (
        train_df.groupby("unit")[sensor]
        .transform(lambda x: x.rolling(window, min_periods=2).std())
    )
```

```python
print("New shape:", train_df.shape)

display(train_df.head())
```

```python
# Count NaN values
nan_counts = train_df.isnull().sum()

print("Total NaN values:", nan_counts.sum())

# Show only columns containing NaN
display(nan_counts[nan_counts > 0])
```

```python
# Feature columns created by us
feature_cols = [
    col for col in train_df.columns
    if "_diff" in col or "_roll_mean" in col or "_roll_std" in col
]

# Fill missing values within each engine
train_df[feature_cols] = (
    train_df.groupby("unit")[feature_cols]
    .transform(lambda x: x.ffill().bfill())
)

# Final safety check
print("Remaining NaN values:", train_df.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

units = train_df["unit"].unique()

train_units, val_units = train_test_split(
    units,
    test_size=0.2,
    random_state=42
)

train_data = train_df[
    train_df["unit"].isin(train_units)
].copy()

val_data = train_df[
    train_df["unit"].isin(val_units)
].copy()

print("Training engines:", len(train_units))
print("Validation engines:", len(val_units))
```

```python
input_features = [
    col for col in train_df.columns
    if col not in ["unit", "cycle", "RUL", "failure_risk"]
]
```

```python
print("Number of input features:", len(input_features))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

train_data[input_features] = scaler.fit_transform(
    train_data[input_features]
)

val_data[input_features] = scaler.transform(
    val_data[input_features]
)
```

```python
WINDOW_SIZE = 30
```

```python
import numpy as np

def create_sequences(df, feature_cols, window_size=30):
    X = []
    y_rul = []
    y_risk = []

    # Process each engine separately
    for unit_id, engine_data in df.groupby("unit"):

        # Sort by cycle
        engine_data = engine_data.sort_values("cycle")

        features = engine_data[feature_cols].values
        rul = engine_data["RUL"].values
        risk = engine_data["failure_risk"].values

        # Create sliding windows
        for i in range(window_size, len(engine_data) + 1):

            X.append(
                features[i-window_size:i]
            )

            # Target corresponds to the LAST cycle in the window
            y_rul.append(
                rul[i-1]
            )

            y_risk.append(
                risk[i-1]
            )

    return (
        np.array(X, dtype=np.float32),
        np.array(y_rul, dtype=np.float32),
        np.array(y_risk, dtype=np.float32)
    )
```

```python
X_train, y_train_rul, y_train_risk = create_sequences(
    train_data,
    input_features,
    window_size=WINDOW_SIZE
)

X_val, y_val_rul, y_val_risk = create_sequences(
    val_data,
    input_features,
    window_size=WINDOW_SIZE
)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
```

```python
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
```

```python
n_features = X_train.shape[2]

inputs = Input(shape=(WINDOW_SIZE, n_features))

x = LSTM(64, return_sequences=False)(inputs)
x = Dropout(0.2)(x)

x = Dense(32, activation="relu")(x)

rul_output = Dense(1, name="rul")(x)

risk_output = Dense(1, activation="sigmoid", name="risk")(x)

model = Model(
    inputs=inputs,
    outputs=[rul_output, risk_output]
)

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss={
        "rul": "mse",
        "risk": "binary_crossentropy"
    },
    metrics={
        "rul": ["mae"],
        "risk": ["accuracy", "AUC"]
    }
)
```

```python
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("y_train_rul:", y_train_rul.shape)
print("y_train_risk:", y_train_risk.shape)
```

```python
model.compile(
    optimizer="adam",
    loss={
        "rul": "mse",
        "risk": "binary_crossentropy"
    },
    metrics={
        "rul": ["mae"],
        "risk": ["accuracy", "AUC"]
    }
)
```

```python
history = model.fit(
    X_train,
    {
        "rul": y_train_rul,
        "risk": y_train_risk
    },
    validation_data=(
        X_val,
        {
            "rul": y_val_rul,
            "risk": y_val_risk
        }
    ),
    epochs=30,
    batch_size=64,
    verbose=1
)
```

```python
import numpy as np
import pandas as pd

# Use the latest 30-cycle window from each validation engine
def calculate_abnormal_score(engine_data, sensor_cols, window_size=30):
    engine_data = engine_data.sort_values("cycle").copy()

    # Take the latest window
    recent = engine_data.tail(window_size)

    scores = []

    for sensor in sensor_cols:
        values = recent[sensor].values

        # Compare recent average with earlier part of the window
        split = len(values) // 2

        early_mean = np.mean(values[:split])
        recent_mean = np.mean(values[split:])

        early_std = np.std(values[:split]) + 1e-8

        # Magnitude of change in standard-deviation units
        change_score = abs(recent_mean - early_mean) / early_std

        scores.append(change_score)

    # Average change across sensors
    abnormal_score = np.mean(scores)

    return abnormal_score
```

```python
def get_abnormal_status(score):

    if score >= 2.0:
        return "ABNORMAL / HIGH DEGRADATION"

    elif score >= 1.0:
        return "WARNING / DEGRADATION DETECTED"

    else:
        return "NORMAL"

print("Abnormal pattern detection function ready.")
```

```python
# Pick one engine from validation data
sample_unit = val_data["unit"].iloc[0]

sample_engine = val_data[
    val_data["unit"] == sample_unit
].copy()

abnormal_score = calculate_abnormal_score(
    sample_engine,
    sensor_cols,
    WINDOW_SIZE
)

status = get_abnormal_status(abnormal_score)

print("========== ABNORMAL PATTERN DETECTION ==========")
print(f"Engine ID        : {sample_unit}")
print(f"Abnormal Score   : {abnormal_score:.2f}")
print(f"Pattern Status   : {status}")
```

```python
def generate_maintenance_alert(predicted_rul, failure_probability, abnormal_score):

    failure_probability = failure_probability * 100

    if (
        predicted_rul <= 10
        or failure_probability >= 80
        or abnormal_score >= 2.0
    ):
        status = "CRITICAL"
        alert = "🚨 IMMEDIATE MAINTENANCE REQUIRED"

    elif (
        predicted_rul <= 30
        or failure_probability >= 50
        or abnormal_score >= 1.0
    ):
        status = "WARNING"
        alert = "⚠️ SCHEDULE MAINTENANCE SOON"

    else:
        status = "HEALTHY"
        alert = "✅ NO IMMEDIATE MAINTENANCE REQUIRED"

    return status, alert
```
